```python
%matplotlib inline
from autoviz import AutoViz_Class
import pandas as pd
```

Importing the csv file

```python
df=pd.read_csv('ml_features.csv')
```

```python
AV=AutoViz_Class()
dft=AV.AutoViz(
    filename="ml_feature.csv",
    sep=",",
    depVar="is_suspicious_tx",
    dfte=df,
    header=0,
    verbose=2,
    chart_format="png",
    max_rows_analyzed=110000,
    max_cols_analyzed=35

)
```
